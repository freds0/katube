```python
import argparse
import glob
import os
from os import makedirs
from os.path import join, exists
import tqdm
import numpy as np
from scipy.io import wavfile
from scipy.interpolate import interp1d

number_bits = 16
encoding = "signed-integer"
number_channels = 1
new_sample_rate = 24000

def downsampling(folder, wav_dir, new_sample_rate, force):
    for wav_path in glob.glob(join(folder, wav_dir) + "/*.wav"):
        original_sample_rate, audio_data = wavfile.read(wav_path)
        
        # Define the new sample rate
        new_sample_rate = 24000
        
        # Create a time array based on the original sample rate
        original_duration = len(audio_data) / original_sample_rate
        original_time = np.linspace(0, original_duration, len(audio_data))
        
        # Create a new time array based on the new sample rate
        new_length = int(original_duration * new_sample_rate)
        new_time = np.linspace(0, original_duration, new_length)
        
        # Perform linear interpolation
        interpolator = interp1d(original_time, audio_data.T, kind='linear')
        new_audio_data = interpolator(new_time).T
        
        # Save the upsampled audio to a new WAV file
        wavfile.write(wav_path, new_sample_rate, new_audio_data.astype(audio_data.dtype))

    return True
```
