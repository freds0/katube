```python
import os

def excluir_pastas_vazias(diretorio):
    for root, dirs, files in os.walk(diretorio, topdown=False):
        for dir in dirs:
            caminho = os.path.join(root, dir)
            # Verifica se o diretório está vazio
            if not os.listdir(caminho):
                os.rmdir(caminho)
                print(f'Diretório vazio removido: {caminho}')

excluir_pastas_vazias(path)
```

```python
base_path = path
# Dicionário para armazenar o número de segmentos por canal
channel_segments_count = {}

# Percorra cada canal na pasta base
for channel in os.listdir(base_path):
    channel_path = os.path.join(base_path, channel)
    time = 0
    # Verifique se o caminho é um diretório (um canal)
    if os.path.isdir(channel_path):
        # Contador de segmentos para o canal atual
        segment_count = 0
        
        # Percorra cada subpasta (segmento) dentro do canal
        for root, dirs, files in os.walk(channel_path):
            # Incrementa o contador para cada arquivo encontrado
            segment_count += len(files)
        
        channel_segments_count[channel] = (segment_count, time)

# Exiba o número de segmentos por canal
for channel, info in channel_segments_count.items():
    print(f'Canal: {channel}: {info[0]} Segmentos ')
```
